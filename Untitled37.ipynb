```python
# ============================================================
# ENTERPRISE MULTI-TENANCY & RBAC
# COMPLETE PYTHON ANALYSIS
# ============================================================

import pandas as pd
import numpy as np


# ============================================================
# 1. LOAD CLEANED CSV
# ============================================================

file_path = "/content/Cleaned_Tenant_Data.csv"

df = pd.read_csv(file_path)

print("=" * 70)
print("TASK 16 - CLEANED DATA LOADED")
print("=" * 70)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())


# ============================================================
# 2. BASIC DATA INFORMATION
# ============================================================

print("\n========== BASIC INFORMATION ==========")

print("Total Records:", len(df))
print("Total Columns:", len(df.columns))

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Info:")
df.info()


# ============================================================
# 3. CONVERT DATA TYPES
# ============================================================

if "event_timestamp" in df.columns:

    df["event_timestamp"] = pd.to_datetime(
        df["event_timestamp"],
        errors="coerce"
    )

numeric_columns = [
    "event_id",
    "event_count",
    "amount"
]

for col in numeric_columns:

    if col in df.columns:

        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )


# ============================================================
# 4. BASIC DATA QUALITY CHECK
# ============================================================

print("\n========== DATA QUALITY CHECK ==========")

print("Total Missing Values:")
print(df.isnull().sum())

print("\nTotal Duplicate Rows:")
print(df.duplicated().sum())


# ============================================================
# 5. DATASET SUMMARY
# ============================================================

print("\n========== DATASET SUMMARY ==========")

print(
    "Unique Tenants:",
    df["tenant_id"].nunique()
)

print(
    "Unique Users:",
    df["user_id"].nunique()
)

print(
    "Unique Events:",
    df["event_id"].nunique()
)

print(
    "Total Event Count:",
    int(df["event_count"].sum())
)

print(
    "Total Amount:",
    round(df["amount"].sum(), 2)
)


# ============================================================
# 6. TENANT OVERVIEW
# ============================================================

tenant_overview = (
    df.groupby(
        ["tenant_id", "tenant_name", "industry"]
    )
    .agg(
        total_records=("event_id", "count"),
        total_events=("event_count", "sum"),
        total_amount=("amount", "sum"),
        unique_users=("user_id", "nunique"),
        successful_events=(
            "status",
            lambda x: (x == "Success").sum()
        ),
        failed_events=(
            "status",
            lambda x: (x == "Failed").sum()
        )
    )
    .reset_index()
)

tenant_overview["success_rate"] = (
    tenant_overview["successful_events"]
    /
    tenant_overview["total_records"]
    * 100
).round(2)

tenant_overview["failure_rate"] = (
    tenant_overview["failed_events"]
    /
    tenant_overview["total_records"]
    * 100
).round(2)

print("\n========== TENANT OVERVIEW ==========")

display(tenant_overview)


# ============================================================
# 7. TENANT-WISE RANKING
# ============================================================

tenant_ranking = tenant_overview.sort_values(
    by="total_amount",
    ascending=False
).reset_index(drop=True)

tenant_ranking["amount_rank"] = (
    tenant_ranking.index + 1
)

print("\n========== TENANT RANKING BY AMOUNT ==========")

display(tenant_ranking)


# ============================================================
# 8. USER-WISE ANALYSIS
# ============================================================

user_analysis = (
    df.groupby(
        ["tenant_id", "tenant_name", "user_id"]
    )
    .agg(
        total_records=("event_id", "count"),
        total_events=("event_count", "sum"),
        total_amount=("amount", "sum"),
        successful_events=(
            "status",
            lambda x: (x == "Success").sum()
        ),
        failed_events=(
            "status",
            lambda x: (x == "Failed").sum()
        )
    )
    .reset_index()
)

print("\n========== USER-WISE ANALYSIS ==========")

display(user_analysis.head(20))


# ============================================================
# 9. EVENT TYPE ANALYSIS
# ============================================================

event_analysis = (
    df.groupby("event_type")
    .agg(
        total_records=("event_id", "count"),
        total_events=("event_count", "sum"),
        total_amount=("amount", "sum"),
        unique_users=("user_id", "nunique"),
        unique_tenants=("tenant_id", "nunique")
    )
    .reset_index()
    .sort_values(
        "total_records",
        ascending=False
    )
)

print("\n========== EVENT TYPE ANALYSIS ==========")

display(event_analysis)


# ============================================================
# 10. TENANT + EVENT TYPE ANALYSIS
# ============================================================

tenant_event_analysis = (
    df.groupby(
        [
            "tenant_id",
            "tenant_name",
            "event_type"
        ]
    )
    .agg(
        total_records=("event_id", "count"),
        total_events=("event_count", "sum"),
        total_amount=("amount", "sum"),
        unique_users=("user_id", "nunique")
    )
    .reset_index()
)

print("\n========== TENANT + EVENT TYPE ANALYSIS ==========")

display(tenant_event_analysis.head(30))


# ============================================================
# 11. REGION-WISE ANALYSIS
# ============================================================

region_analysis = (
    df.groupby("region")
    .agg(
        total_records=("event_id", "count"),
        total_events=("event_count", "sum"),
        total_amount=("amount", "sum"),
        unique_tenants=("tenant_id", "nunique"),
        unique_users=("user_id", "nunique")
    )
    .reset_index()
    .sort_values(
        "total_amount",
        ascending=False
    )
)

print("\n========== REGION-WISE ANALYSIS ==========")

display(region_analysis)


# ============================================================
# 12. TENANT + REGION ANALYSIS
# ============================================================

tenant_region_analysis = (
    df.groupby(
        [
            "tenant_id",
            "tenant_name",
            "region"
        ]
    )
    .agg(
        total_records=("event_id", "count"),
        total_events=("event_count", "sum"),
        total_amount=("amount", "sum")
    )
    .reset_index()
)

print("\n========== TENANT + REGION ANALYSIS ==========")

display(tenant_region_analysis.head(30))


# ============================================================
# 13. STATUS ANALYSIS
# ============================================================

status_analysis = (
    df.groupby("status")
    .agg(
        total_records=("event_id", "count"),
        total_events=("event_count", "sum"),
        total_amount=("amount", "sum"),
        unique_tenants=("tenant_id", "nunique")
    )
    .reset_index()
)

status_analysis["percentage"] = (
    status_analysis["total_records"]
    /
    status_analysis["total_records"].sum()
    * 100
).round(2)

print("\n========== STATUS ANALYSIS ==========")

display(status_analysis)


# ============================================================
# 14. INDUSTRY-WISE ANALYSIS
# ============================================================

industry_analysis = (
    df.groupby("industry")
    .agg(
        total_records=("event_id", "count"),
        total_events=("event_count", "sum"),
        total_amount=("amount", "sum"),
        unique_tenants=("tenant_id", "nunique"),
        unique_users=("user_id", "nunique")
    )
    .reset_index()
    .sort_values(
        "total_amount",
        ascending=False
    )
)

print("\n========== INDUSTRY-WISE ANALYSIS ==========")

display(industry_analysis)


# ============================================================
# 15. MONTHLY ANALYSIS
# ============================================================

df["event_month"] = (
    df["event_timestamp"]
    .dt.to_period("M")
    .astype(str)
)

monthly_analysis = (
    df.groupby("event_month")
    .agg(
        total_records=("event_id", "count"),
        total_events=("event_count", "sum"),
        total_amount=("amount", "sum"),
        unique_tenants=("tenant_id", "nunique"),
        unique_users=("user_id", "nunique")
    )
    .reset_index()
)

print("\n========== MONTHLY ANALYSIS ==========")

display(monthly_analysis)


# ============================================================
# 16. TENANT MONTHLY ANALYSIS
# ============================================================

tenant_monthly_analysis = (
    df.groupby(
        [
            "event_month",
            "tenant_id",
            "tenant_name"
        ]
    )
    .agg(
        total_records=("event_id", "count"),
        total_events=("event_count", "sum"),
        total_amount=("amount", "sum")
    )
    .reset_index()
)

print("\n========== TENANT MONTHLY ANALYSIS ==========")

display(tenant_monthly_analysis.head(30))


# ============================================================
# 17. SUCCESS / FAILURE BY TENANT
# ============================================================

tenant_status_analysis = (
    df.groupby(
        [
            "tenant_id",
            "tenant_name",
            "status"
        ]
    )
    .agg(
        total_records=("event_id", "count"),
        total_events=("event_count", "sum"),
        total_amount=("amount", "sum")
    )
    .reset_index()
)

print("\n========== TENANT STATUS ANALYSIS ==========")

display(tenant_status_analysis)


# ============================================================
# 18. TENANT CONSISTENCY CHECK
# ============================================================

tenant_name_check = (
    df.groupby("tenant_id")["tenant_name"]
    .nunique()
)

inconsistent_tenant_names = (
    tenant_name_check[
        tenant_name_check > 1
    ]
)

print("\n========== TENANT NAME CONSISTENCY ==========")

if len(inconsistent_tenant_names) == 0:

    print(
        "PASS - Every Tenant ID has one Tenant Name."
    )

else:

    print(
        "FAIL - Inconsistent Tenant Names found."
    )

    print(inconsistent_tenant_names)


# ============================================================
# 19. TENANT / INDUSTRY CONSISTENCY CHECK
# ============================================================

tenant_industry_check = (
    df.groupby("tenant_id")["industry"]
    .nunique()
)

inconsistent_industries = (
    tenant_industry_check[
        tenant_industry_check > 1
    ]
)

print("\n========== TENANT INDUSTRY CONSISTENCY ==========")

if len(inconsistent_industries) == 0:

    print(
        "PASS - Tenant to Industry mapping is consistent."
    )

else:

    print(
        "FAIL - Inconsistent Industry mapping found."
    )

    print(inconsistent_industries)


# ============================================================
# 20. USER TENANT ISOLATION CHECK
# ============================================================

user_tenant_check = (
    df.groupby("user_id")["tenant_id"]
    .nunique()
)

cross_tenant_users = (
    user_tenant_check[
        user_tenant_check > 1
    ]
)

print("\n========== USER TENANT ISOLATION ==========")

if len(cross_tenant_users) == 0:

    print(
        "PASS - No user belongs to multiple tenants."
    )

else:

    print(
        "FAIL - Potential cross-tenant users found."
    )

    print(cross_tenant_users)


# ============================================================
# 21. EVENT TENANT ISOLATION CHECK
# ============================================================

event_tenant_check = (
    df.groupby("event_id")["tenant_id"]
    .nunique()
)

cross_tenant_events = (
    event_tenant_check[
        event_tenant_check > 1
    ]
)

print("\n========== EVENT TENANT ISOLATION ==========")

if len(cross_tenant_events) == 0:

    print(
        "PASS - Every Event ID belongs to one tenant only."
    )

else:

    print(
        "FAIL - Potential cross-tenant events found."
    )

    print(cross_tenant_events)


# ============================================================
# 22. DUPLICATE EVENT ID CHECK
# ============================================================

duplicate_event_ids = (
    df["event_id"].duplicated().sum()
)

print("\n========== DUPLICATE EVENT ID CHECK ==========")

print(
    "Duplicate Event IDs:",
    duplicate_event_ids
)


# ============================================================
# 23. CROSS-TENANT LEAKAGE VALIDATION
# ============================================================

print("\n========== CROSS-TENANT DATA LEAKAGE ==========")

leakage_checks = []

# Check Event ID
if len(cross_tenant_events) == 0:

    leakage_checks.append(
        ["Event ID Isolation", "PASS"]
    )

else:

    leakage_checks.append(
        ["Event ID Isolation", "FAIL"]
    )


# Check User ID
if len(cross_tenant_users) == 0:

    leakage_checks.append(
        ["User ID Isolation", "PASS"]
    )

else:

    leakage_checks.append(
        ["User ID Isolation", "FAIL"]
    )


# Check Tenant Name
if len(inconsistent_tenant_names) == 0:

    leakage_checks.append(
        ["Tenant Name Consistency", "PASS"]
    )

else:

    leakage_checks.append(
        ["Tenant Name Consistency", "FAIL"]
    )


leakage_report = pd.DataFrame(
    leakage_checks,
    columns=[
        "Check",
        "Result"
    ]
)

display(leakage_report)


# ============================================================
# 24. AGGREGATE BENCHMARK
# ============================================================

number_of_tenants = df["tenant_id"].nunique()

benchmark_analysis = pd.DataFrame({

    "Metric": [
        "Total Tenants",
        "Average Records per Tenant",
        "Average Events per Tenant",
        "Average Amount per Tenant",
        "Median Records per Tenant",
        "Median Amount per Tenant"
    ],

    "Value": [

        number_of_tenants,

        round(
            tenant_overview["total_records"].mean(),
            2
        ),

        round(
            tenant_overview["total_events"].mean(),
            2
        ),

        round(
            tenant_overview["total_amount"].mean(),
            2
        ),

        round(
            tenant_overview["total_records"].median(),
            2
        ),

        round(
            tenant_overview["total_amount"].median(),
            2
        )
    ]
})

print("\n========== AGGREGATE BENCHMARK ==========")

display(benchmark_analysis)


# ============================================================
# 25. TOP TENANT ANALYSIS
# ============================================================

top_tenant = tenant_overview.loc[
    tenant_overview["total_amount"].idxmax()
]

print("\n========== TOP TENANT ==========")

print(
    "Top Tenant:",
    top_tenant["tenant_name"]
)

print(
    "Tenant ID:",
    top_tenant["tenant_id"]
)

print(
    "Total Amount:",
    round(
        top_tenant["total_amount"],
        2
    )
)


# ============================================================
# 26. MOST ACTIVE TENANT
# ============================================================

most_active_tenant = tenant_overview.loc[
    tenant_overview["total_events"].idxmax()
]

print("\n========== MOST ACTIVE TENANT ==========")

print(
    "Tenant:",
    most_active_tenant["tenant_name"]
)

print(
    "Total Events:",
    int(
        most_active_tenant["total_events"]
    )
)


# ============================================================
# 27. HIGHEST FAILURE TENANT
# ============================================================

highest_failure_tenant = tenant_overview.loc[
    tenant_overview["failure_rate"].idxmax()
]

print("\n========== HIGHEST FAILURE RATE TENANT ==========")

print(
    "Tenant:",
    highest_failure_tenant["tenant_name"]
)

print(
    "Failure Rate:",
    highest_failure_tenant["failure_rate"],
    "%"
)


# ============================================================
# 28. FINAL DATA QUALITY REPORT
# ============================================================

quality_report = pd.DataFrame({

    "Column": df.columns,

    "Data_Type": [
        str(df[col].dtype)
        for col in df.columns
    ],

    "Missing_Values": [
        df[col].isnull().sum()
        for col in df.columns
    ],

    "Unique_Values": [
        df[col].nunique()
        for col in df.columns
    ]
})

print("\n========== FINAL DATA QUALITY REPORT ==========")

display(quality_report)


# ============================================================
# 29. FINAL VALIDATION SUMMARY
# ============================================================

validation_summary = pd.DataFrame({

    "Validation_Check": [

        "Duplicate Rows",
        "Missing Values",
        "Duplicate Event IDs",
        "Cross-Tenant Users",
        "Cross-Tenant Events",
        "Inconsistent Tenant Names",
        "Inconsistent Industries"
    ],

    "Result": [

        df.duplicated().sum(),

        df.isnull().sum().sum(),

        duplicate_event_ids,

        len(cross_tenant_users),

        len(cross_tenant_events),

        len(inconsistent_tenant_names),

        len(inconsistent_industries)
    ]
})

print("\n========== VALIDATION SUMMARY ==========")

display(validation_summary)


# ============================================================
# 30. FINAL DATASET STATISTICS
# ============================================================

final_statistics = pd.DataFrame({

    "Metric": [

        "Total Records",
        "Total Columns",
        "Unique Tenants",
        "Unique Users",
        "Unique Events",
        "Total Event Count",
        "Total Amount"
    ],

    "Value": [

        len(df),

        len(df.columns),

        df["tenant_id"].nunique(),

        df["user_id"].nunique(),

        df["event_id"].nunique(),

        int(df["event_count"].sum()),

        round(df["amount"].sum(), 2)
    ]
})

print("\n========== FINAL STATISTICS ==========")

display(final_statistics)


# ============================================================
# 31. SAVE COMPLETE ANALYSIS TO EXCEL
# ============================================================

output_excel = (
    "/content/Task_16_Python_Analysis_Output.xlsx"
)

with pd.ExcelWriter(
    output_excel,
    engine="openpyxl"
) as writer:

    df.to_excel(
        writer,
        index=False,
        sheet_name="Cleaned_Data"
    )

    tenant_overview.to_excel(
        writer,
        index=False,
        sheet_name="Tenant_Overview"
    )

    tenant_ranking.to_excel(
        writer,
        index=False,
        sheet_name="Tenant_Ranking"
    )

    user_analysis.to_excel(
        writer,
        index=False,
        sheet_name="User_Analysis"
    )

    event_analysis.to_excel(
        writer,
        index=False,
        sheet_name="Event_Analysis"
    )

    tenant_event_analysis.to_excel(
        writer,
        index=False,
        sheet_name="Tenant_Event"
    )

    region_analysis.to_excel(
        writer,
        index=False,
        sheet_name="Region_Analysis"
    )

    tenant_region_analysis.to_excel(
        writer,
        index=False,
        sheet_name="Tenant_Region"
    )

    status_analysis.to_excel(
        writer,
        index=False,
        sheet_name="Status_Analysis"
    )

    industry_analysis.to_excel(
        writer,
        index=False,
        sheet_name="Industry_Analysis"
    )

    monthly_analysis.to_excel(
        writer,
        index=False,
        sheet_name="Monthly_Analysis"
    )

    tenant_monthly_analysis.to_excel(
        writer,
        index=False,
        sheet_name="Tenant_Monthly"
    )

    tenant_status_analysis.to_excel(
        writer,
        index=False,
        sheet_name="Tenant_Status"
    )

    benchmark_analysis.to_excel(
        writer,
        index=False,
        sheet_name="Benchmark"
    )

    leakage_report.to_excel(
        writer,
        index=False,
        sheet_name="Leakage_Check"
    )

    quality_report.to_excel(
        writer,
        index=False,
        sheet_name="Quality_Report"
    )

    validation_summary.to_excel(
        writer,
        index=False,
        sheet_name="Validation"
    )

    final_statistics.to_excel(
        writer,
        index=False,
        sheet_name="Final_Statistics"
    )


# ============================================================
# 32. SAVE FINAL ANALYSIS CSV
# ============================================================

output_csv = (
    "/content/Task_16_Final_Analysis.csv"
)

df.to_csv(
    output_csv,
    index=False
)


# ============================================================
# 33. FINAL RESULT
# ============================================================

print("\n")
print("=" * 70)
print("TASK 16 PYTHON ANALYSIS COMPLETED")
print("=" * 70)

print("\nFinal Records:", len(df))
print("Final Columns:", len(df.columns))

print(
    "Unique Tenants:",
    df["tenant_id"].nunique()
)

print(
    "Unique Users:",
    df["user_id"].nunique()
)

print(
    "Unique Events:",
    df["event_id"].nunique()
)

print(
    "Total Events:",
    int(df["event_count"].sum())
)

print(
    "Total Amount:",
    round(df["amount"].sum(), 2)
)

print(
    "Duplicate Rows:",
    df.duplicated().sum()
)

print(
    "Missing Values:",
    df.isnull().sum().sum()
)

print(
    "Cross-Tenant Users:",
    len(cross_tenant_users)
)

print(
    "Cross-Tenant Events:",
    len(cross_tenant_events)
)

print("\nExcel Analysis File:")
print(output_excel)

print("\nFinal CSV File:")
print(output_csv)

print("\nNO CHARTS CREATED")
print("NO DASHBOARD CREATED")
print("TABLEAU WILL BE USED FOR VISUALIZATION")

print("\nTASK 16 COMPLETE!")
```
